#### 2 Tracé de graphiques simples avec Matplotlib

```python
# Generation de données
x=np.linspace(0,10,1000)
y=np.exp(x)
```

```python
# Tracer un graphique simple
plt.plot(x,y)
# Sur d'autres environnements, on aura besoin de taper (plt.show()) pour afficher le graphique
```

### 3. Personnalisation de graphique avec Matplotlib

```python
# Changer la taille d'un graphique avec la largeur et la hauteur de la figure en pouce
plt.figure() # plt.figure(figsize=(5,5))
plt.plot(x,y)
```

```python
# Titre et axes
plt.plot(x,y)
plt.title("Fonction exponentielle")
plt.xlabel("x")
plt.ylabel("valeur")
plt.show()
```

```python
# Changement de couleur
plt.plot(x,y,c="red")
plt.title("Fonction exponentielle")
plt.xlabel("x")
plt.ylabel("valeur")
plt.show()
```

```python
# Changer le type de ligne
plt.plot(x,y,c="red",ls="--")
plt.title("Fonction exponentielle")
plt.xlabel("x")
plt.ylabel("valeur")
plt.show()
```

```python
# Ajouter une legende

plt.plot(x,y,c="red",ls="--", label="exp")
plt.title("Fonction exponentielle")
plt.xlabel("x")
plt.ylabel("valeur")
plt.legend()
plt.show()
```

```python
# Ajouter une grille sur le graphique
plt.plot(x,y,c="red",ls="--", label="exp")
plt.title("Fonction exponentielle")
plt.xlabel("x")
plt.ylabel("valeur")
plt.legend()
plt.grid()
plt.show()
```

### 4. Faire plusieurs graphiques

#### 4.1 Faire des graphiques sur un meme carré

```python
# Creer un graphique vide

z=2*y
y3=z*2
plt.plot(x,y,c="red",ls="-", label="exp")
plt.plot(x,z,c="blue",ls="-", label="2*exp")
plt.plot(x,y3,c="green",ls="-", label="4*exp")
plt.title("Fonction exponentielle")
plt.xlabel("x")
plt.ylabel("valeur")
plt.legend()
plt.grid()
plt.show()
```

```python
# Pour zoomer une partie du graphique (par exemple visionner sur (6,10))
# On va changer les limites des axes

z=2*y
y3=z*2
plt.plot(x,y,c="red",ls="-", label="exp")
plt.plot(x,z,c="blue",ls="-", label="2*exp")
plt.plot(x,y3,c="green",ls="-", label="4*exp")
plt.xlim((6,10))
plt.ylim((1000,80000))
plt.title("Fonction exponentielle")
plt.xlabel("x")
plt.ylabel("valeur")
plt.legend()
plt.grid()
plt.show()
```

#### 4.2 Faire des subplots

#### 4.2.1 Approche fonctionnelle

```python
# Graphique avec subplot (Approche fonctionnelle)
plt.subplot(2,2,1) # 2: nbre de lignes, 1: nbre de colonnes et 1: la position du graphique 
plt.plot(x,y,c="red",ls="-", label="exp")
plt.subplot(2,2,2)
plt.plot(x,y,c="blue",ls="-", label="exp")
plt.subplot(2,2,3)
plt.plot(x,y,c="green",ls="-", label="exp")
plt.subplot(2,2,4)
plt.plot(x,y,c="black",ls="-", label="exp")
plt.subplots_adjust(wspace=0.5)
```

#### 4.2.2  Approche orientée objet


```python
# Subplots
fig,axes=plt.subplots(2,1) # 2 lignes et 1 colonne
axes[0].plot(x,y,c="red")
axes[1].plot(x,y,c="blue")
```

```python
# Sauvegarder son graphique dans un fichier
plt.subplot(2,2,1) 
plt.plot(x,y,c="red",ls="-", label="exp")
plt.subplot(2,2,2)
plt.plot(x,y,c="blue",ls="-", label="exp")
plt.subplot(2,2,3)
plt.plot(x,y,c="green",ls="-", label="exp")
plt.subplot(2,2,4)
plt.plot(x,y,c="black",ls="-", label="exp")
plt.subplots_adjust(wspace=0.5)
plt.savefig("figure.png")
plt.savefig("figure.jpeg")
```

#### 3.5 Heatmap 

```python
# heatmap : example sur la matrice de correlation
# Sélectionner uniquement les colonnes numériques
colonnes_numeriques = titanic.select_dtypes(include=['number'])
# Calculer la matrice de corrélation
matrice_correlation = colonnes_numeriques.corr()
matrice_correlation

# heatmap
sns.heatmap(matrice_correlation,cmap="YlGnBu",annot=True) # Possibilité d'enlever la palette de couleur (cpam) annot rajoute le rapport de corre sur le graphique
```

```python
# Plus c'est bleu, plus la correlation entre les deux variables est importante
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# Supposons que 'titanic' est un DataFrame pandas contenant les données du Titanic
# Sélectionner uniquement les colonnes numériques
corr_matrix = titanic.select_dtypes(include=['number']).corr()

# Créer la heatmap avec matplotlib
plt.figure(figsize=(6, 5))
im = plt.matshow(corr_matrix, cmap='coolwarm', fignum=1)

# Ajouter les valeurs de corrélation sur chaque cellule
for i in range(corr_matrix.shape[0]):
    for j in range(corr_matrix.shape[1]):
        text = plt.text(j, i, round(corr_matrix.iloc[i, j], 2),
                        ha="center", va="center", color="black")

# Ajouter la barre de couleur
cbar = plt.colorbar(im)

# Modifier les ticks des axes et les labels
ax = plt.gca()
ax.set_xticks(np.arange(corr_matrix.shape[1]))
ax.set_xticklabels(corr_matrix.columns, rotation=45)
ax.set_yticks(np.arange(corr_matrix.shape[0]))
ax.set_yticklabels(corr_matrix.columns)

# Afficher le graphique
plt.title("Matrice de corrélation des variables numériques du Titanic")
plt.show()
```

#### 4 . Plotly (autre package interessant, tres utilisé en matiere de visualisation)

```python
import plotly.express as px
# Creer un graphique interessant interactif
fig = px.scatter(
    iris,
    x="sepal_length",
    y="sepal_width",
    color="species",
    size="petal_length",
    title="Iris Dataset: Sepal Length vs Sepal Width",
    labels={"sepal_length": "Longueur Sepal (cm)", "sepal_width": "Largeur Sepal (cm)"},
)
fig.show()
```

```python
# Charger le dataset flights
flights = sns.load_dataset("flights")

# Creer un graphique interactif
fig = px.line(
    flights,
    x= "year",
    y="passengers",
    color="month",
    title="Evolution du nombre de passagers par mois (1949 - 1969)",
    labels={
        "year":"Année",
        "passengers":"Nombre de passagers",
        "month": "Mois",
    },
)
# Affichage du graphique interactif
fig.show()
```

```python
# Autre exemple
# Charger le dataset gapminder
gapminder = px.data.gapminder()
# creer un graphique interactif
fig = px.scatter(
    gapminder,
    x="gdpPercap",
    y = "lifeExp",
    size ="pop",
    color="continent",
    hover_name="country",
    log_x=True,
    range_y=[20, 90],
    animation_frame="year",
    title="Evolution de l'espérane de vie et du PIB par habitant (1952 - 2007)",
    labels={
        "gdpPercap":"PIB par habitant (log)",
        "lifeExp": "Esperance de vie",
        "year":"Année",
        "pop":"Population",
        "continent":"Continent",
    },
    height=600)

# Affiner l'apparence du graphique
fig.update_traces(marker=dict(line=dict(width=1, color="rgba(0, 0, 0, 0.5)")))
fig.update_layout(showlegend=True, legend=dict(orientation="h", yanchor="bottom", y=-0.2, xanchor="center", x=0.5))

# Affichage du graphique interactif
fig.show()

    
```
