```python
import requests
import os
```

```python
# Creation d'un dossier pour stocker les fichiers téléchargés
os.makedirs("yellow_taxi_data", exist_ok=True)
```

```python
# Creation de la fonction d'automatisation:
for year in range(2025, 2012, -1):
    for month in range(1, 13):
        # Construction de l'url du téléchargement en fonction de l'année et du mois 
        download_url = f"https://d37ci6vzurychx.cloudfront.net/trip-data/yellow_tripdata_{year}-{month}.parquet"
        # print(download_url)
        file_name = f"yellow_taxi_data/ytripdata_{year}-{month:02d}.parquet"
        # print(file_name)

        # Télecharger le fichier 
        print(f"Téléchargement de {file_name}...")
        response = requests.get(download_url)

        if response.status_code == 200: # Rappel: quand une requete a marché, le statut c'est 200)
            with open(file_name, "wb") as f:
                f.write(response.content)
        else:
            print(f"Impossible de télécharger {file_name}. Code de statut HTTP: {response.status_code}")

print("Télechargement terminé!")
```

```python
# les données seront ajoutés tous les mois avec un retard de deux mois

from datetime import datetime, timedelta
```

```python
# datetime.today()
# Fonction pour obtenir la date actuelle moins 2 mois
def get_previous_month():
    today = datetime.today()
    two_months_ago = today - timedelta(days=60)
    return two_months_ago.strftime("%Y - %m")

# get_previous_month()

# Année courante
current_year = datetime.now().year
# current_year
current_month = datetime.now().month

# Previous month
previous_month = get_previous_month() 

# Mise à jour du code pour actualiser les données en cas d'ajout de données

# Creation de la fonction d'automatisation:
for year in range(2025, current_year + 1):
    for month in range(1, 13):
        # Ignorer les mois antérieurs à la date actuelle moins deux mois
        if year == current_year and month > datetime.now().month:
            continue  
        if f"{year}-{month:02d}" < previous_month:
            continue
        # Construction de l'url du téléchargement en fonction de l'année et du mois 
        download_url = f"https://d37ci6vzurychx.cloudfront.net/trip-data/yellow_tripdata_{year}-{month}.parquet"
        # print(download_url)
        file_name = f"yellow_taxi_data/ytripdata_{year}-{month:02d}.parquet"
        # print(file_name)

        # Télecharger le fichier 
        print(f"Téléchargement de {file_name}...")
        response = requests.get(download_url)

        if response.status_code == 200: # Rappel: quand une requete a marché, le statut c'est 200)
            with open(file_name, "wb") as f:
                f.write(response.content)
        else:
            print(f"Impossible de télécharger {file_name}. Code de statut HTTP: {response.status_code}")

print("Télechargement terminé!")
```
