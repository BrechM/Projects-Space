```python
# Importation des modules
from load_data import load_data # Si j'ai plussieurs fonction de mon module load_data, on importait le tout en tapant import load_data as l ou from load_data import *
from data_split import data_split
from model_fit import model_fit 
from cross_validation import cross_validate
from model_save import model_save 
```

```python
# Chargement des données
x, y = load_data()
```

```python
y
```

```python
# Données d'entrainemet et de test
x_train, x_test, y_train, y_test = data_split(x, y, test_size=0.2)
```

```python
# Formation du modèle
fitted_model = model_fit(x_train, y_train, max_iter = 100)
```

```python
type(fitted_model)
```

```python
fitted_model.score(x_test, y_test)
```

```python
# Validation croisée
scores = cross_validate(fitted_model, x_test, y_test)
```

```python
scores
```

```python
scores.mean()
```

```python
# Sauvegarde du modele
model_save(fitted_model, scores, seuil=0.9)
```
