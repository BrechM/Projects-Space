## Exemple de code monolithique

```python
from sklearn.datasets import load_breast_cancer
from sklearn.linear_model import LogisticRegression
```

```python
# Chargement des données 
data = load_breast_cancer()
x, y = data['data'], data['target']
```

```python
x
```

```python
y
```

```python
# Pretraitement
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
x = scaler.fit_transform(x)
```

```python
import numpy as np 
np.bincount(y) # Pour avoir une idée de la distribution des données de y, on a 212 pour la modalité 0 et 
# 357 pour la modalité 1
```

```python
# ou bien 
import pandas as pd 
pd.DataFrame(y).value_counts()
```

```python
# Pour determiner les proportions de chaque classe
pd.DataFrame(y).value_counts(normalize=True)
```

On a 62,74 % pour la classe 1 et 37 % pour la classe 0.

Il y'a un déséquilbre de classe donc il faut mettre l'argument class_weight = 'balanced' au niveau de l'algorithme de regression logistique.

```python
# Formation du modele 
clf = LogisticRegression(
    random_state=0,
    max_iter = 500,
    class_weight = 'balanced'
)

clf.fit(x, y)
```

```python
# Score de mon modele
clf.score(x, y)
```

```python
# Validation croisée
from sklearn.model_selection import cross_val_score
scores = cross_val_score(
    clf,
    x,
    y,
    cv=5
)

print(scores)
print("Score moyen :", scores.mean()) # donne une estimation plus robuste et réaliste de la 
# performance du modèle qu’un simple score sur un train/test split.
print("Ecart-type :", scores.std())
```

Le score de mon modele n'est pas loin du score moyen de la validation croisée, on peut donc dire que le modele va bien se généraliser sur de nouvelles données
