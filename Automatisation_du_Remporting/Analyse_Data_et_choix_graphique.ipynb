```python
# On peut agrandir
plt.figure(figsize=(12, 12))

# On va creer la matrice de coorelation
mask = np.triu(np.ones_like(vin_rouge.corr(), dtype=bool))
sns.heatmap(vin_rouge.corr(), mask=mask, center=0, cmap='RdBu', linewidths=1, annot=True, fmt=".2f", vmin=-1, vmax=1)
plt.title('Coefficient de corrélation de Pearson entre variables', fontsize=15, fontweight="bold")
plt.show()
#plt.savefig('red_wine_heatmap.pdf', orientation='landscape')
```

```python
# On peut ensuite tracer des boxplot
# Cette fonction va me sauvegarder automatiquement les boxplots en pdf

def save_qualityvsfeats_fig(data):
    feats = data.select_dtypes(include = ['float64']).columns.to_list() # on tient compte que des float
    for col in feats:
        Pearson_coef, P_value = st.pearsonr(data[col], data["quality"])
        aff = 'Coef.Pearson =' + str(round(Pearson_coef, 3)) + ', P-valeur = ' + str(round(P_value, 3))
        sns.boxplot(x=data['quality'], y=data[col])
        plt.xlabel('Score de qualité du vin', fontweight="bold")
        plt.ylabel(col, fontweight="bold")
        plt.title('quality VS' + col + '.' + aff, fontsize=15, fontweight="bold")
        plt.show()
        #plt.savefig(col,'VS quality.pdf', orientation='landscape') # On decoche lorsqu'il s'agira de faire du reporting
        plt.close()
```

```python
save_qualityvsfeats_fig(vin_rouge)
```
